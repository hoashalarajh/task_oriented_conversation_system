### Identifying number of unique words in each interaction

```python
def get_unique_wordcount(file_path):
    # mentioning the path
    #file_path = "E://#Journal/Actual_Person_user_study/user_response_files/"
    # getting the ist of files in the directory
    user_responses = os.listdir(file_path)

    # measuring the number of words during each user response
    num_of_uq_words = []

    # reading file by file
    for response_file in user_responses:
        #print (response_file)
        # defining the pattern i need
        pattern01 = "Type your response here:"

        # defininfg word list
        word_counter = []

        # reading the text file
        file1 = open(file_path + response_file,"r")
        for line in file1:
            if pattern01 in line:
                word_counter.append(line[26:])
            else:
                pass
        num_uq_words_set = set(word_counter)
        num_of_uq_words.append(len(num_uq_words_set))
        word_counter = []
        file1.close()
    return (num_of_uq_words)
```
