```python
# importing necessary dependency
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy import stats as st
import os
```

```python
# mentioning the path
file_path = "E://#Journal/Actual_Person_user_study/user_response_files/"
# getting the ist of files in the directory
user_responses = os.listdir(file_path)

# defining a list to store the number of interactions happened
num_interactions_list = []

# reading file by file
for response_file in user_responses:
    print (response_file)
    # defining the pattern i need
    pattern = "Type your response here:"
    # defining the pattern counter
    patt_counter = 0

    # reading the text file
    file1 = open(file_path + response_file,"r")
    for line in file1:
        if pattern in line:
            patt_counter = patt_counter + 1

    # printting the number of interactions
    print (f"The numer of interactions had: {patt_counter}")
    # appending the number of interactions list
    num_interactions_list.append(patt_counter)
    # resetting the pattern counter
    patt_counter = 0
    file1.close()
```

### Getting interaction mode changes

```python
def get_imc(file_path):
    # getting the ist of files in the directory
    user_responses = os.listdir(file_path)

    # defining a list to store the number of interactions happened
    num_imc_list_rand = np.zeros(21)

    user_counter = 0

    # reading file by file
    for response_file in user_responses:
        print (response_file)
        # defining the pattern i need : my search sentence
        pattern = "Randomly selected number:"
        # defining the pattern counter
        patt_counter = 0

        # reading the text file
        file1 = open(file_path + response_file,"r")
        for line in file1:
            if pattern in line:
                patt_counter = patt_counter + 1

        # printting the number of interactions
        print (f"The numer of mode changes: {patt_counter}")
        # appending the number of interactions list
        #num_imc_list_rand.append(patt_counter)
        num_imc_list_rand[user_counter] = patt_counter
        # incrementing user counter
        user_counter = user_counter + 1
        # resetting the pattern counter
        patt_counter = 0
        file1.close()
    # returning the imc
    return (num_imc_list_rand)
```
